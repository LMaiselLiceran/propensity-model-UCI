# 03. Models

In this notebook we will test different model variants.

As always we start by loading the data and preparing it appropriately.

```python
import sys
sys.path.append('..')

import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import TimeSeriesSplit, cross_validate

from src.dataset import load_clients, add_year_block, time_split
from src.features import TARGET, prepare
from src.models import build_pipeline

df = add_year_block(load_clients())
train, test = time_split(df)
X_train, y_train = prepare(train), train[TARGET]
```

## Splitting the data

We split the test data into five different folds.
Fold 1 means train on chunk 1 and test on chunk 2, fold 2 means train on chunks 1-2 and test on 3, fold 3 means train on 1-3 and test on 4, and so on.
The chunks are in chronological order, so we are never using the future to predict the past.

```python
cv = TimeSeriesSplit(n_splits=5)
for i, (tr, te) in enumerate(cv.split(X_train), start=1):
    print(i, len(tr), len(te), y_train.iloc[te].mean().round(3))  # First look at the folds
```

## Logistic regression

We now look at the ROC AUC and PR AUC scores after each fold using a logistic regression.

```python
scoring = {"ROC_AUC": "roc_auc", "PR_AUC": "average_precision"}

logreg = build_pipeline(LogisticRegression(max_iter=1000), include_macro=True)
res = cross_validate(logreg, X_train, y_train, cv=cv, scoring=scoring)
tb = pd.DataFrame(res)[["test_ROC_AUC", "test_PR_AUC"]].round(3)
tb["base_rate"] = [y_train.iloc[te].mean().round(3) for _, te in cv.split(X_train)]
tb["PR_AUC / base"] = round(tb["test_PR_AUC"] / tb["base_rate"], 3)
tb
```

Folds 1-4 show essentially no skill, only after fold 4 can we talk about that.
As the increase actually goes down between the 2nd and 4th folds, it could be that in the beginning patterns were learned that were later reversed.
The average ROC AUC score is about 0.54, not much better than random guessing.

To diagnose this a bit, let us look at the ROC AUC score when we use the trained model back on the trained data.

```python
from sklearn.metrics import roc_auc_score

logreg.fit(X_train, y_train)
print(roc_auc_score(y_train, logreg.predict_proba(X_train)[:, 1]))
```

A ROC AUC score of 0.68 is not bad (it's above the 0.5 threshold), but still modest.
It could be that a linear model is not enough here because the link between features and subscribing differs from period to period.
We will later see if gradient boosting fixes this.

Another possibility comes from the fact that in each fold we use the trained model on months that didn't appear in the training.
This may be unfair because in the final test we will have trained on all months.
We can check how many unseen months there are in the test data of each fold:

```python
for i, (tr, te) in enumerate(cv.split(X_train), start=1):
    seen = set(train.iloc[tr]["month"])
    unseen = (~train.iloc[te]["month"].isin(seen)).mean().round(2)
    print(i, "share of scoring rows with a month not seen in training:", unseen)
```

In all cases, more than 50% of the scoring rows come from months that were not seen in training.
This means that the training folds are actually harder than the actual test, at least in this specific way.

## Shuffled cross-validation

Finally, let us try a shuffled cross-validation.
We train on the same data but shuffle the rows, i.e., we mix all time periods together.

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
res_random = cross_validate(logreg, X_train, y_train, cv=skf, scoring=scoring)
print(res_random["test_ROC_AUC"].round(3), res_random["test_ROC_AUC"].mean().round(3))
print(res_random["test_PR_AUC"].round(3), "base rate:", y_train.mean().round(3))
```

We get a score of 0.67, almost the same as what we got when scoring on the training data a few cells above.
This means that, when time is not being taken into account, the model does as well on data it hasn't seen as on data it has seen.
However, once we train on a period and test on the following one, the score drops to 0.54 on average.

When we mix periods, the model learns in part which periods are good (i.e., the macro conditions and months) and in part something about the actual clients. To separate the two, let us run the model with shuffled rows but this time excluding the macros:

```python
logreg_nm = build_pipeline(LogisticRegression(max_iter=1000), include_macro=False)
r = cross_validate(logreg_nm, X_train, y_train, cv=skf, scoring=scoring)
print(f'Shuffled, no macros: ROC AUC {r["test_ROC_AUC"].mean():.3f}, PR AUC {r["test_PR_AUC"].mean():.3f}')
```

The ROC AUC score barely fell from the 0.67-0.68 marks above.
This excludes the possibility that much of the learned information was related to the time period.
Instead, it's telling us that the skill does come from client and contact features, but these features change over time.

## Gradient boosting

We now test a more complicated model. `HistGradientBoostingClassifier` builds many small decision trees, each correcting the mistakes of previous ones. We want to compare the gradient boosting method with the logistic regression, both in the presence and absence of macro data.

The boosting settings have the following meaning:
- `max_depth = 3`: each tree asks at most three questions in a row, so it can't build overly specific rules.
- `learning_rate = 0.05`: each tree corrects the previous ones in smaller steps.
- `min_samples_leaf = 200`: the final group of clients always contains at least 200 rows, so the model can't fit small details.
- `max_iter = 150`: the number of trees.

```python
def cv_table(model):
    res = cross_validate(model, X_train, y_train, cv=cv, scoring=scoring)
    table = pd.DataFrame(res)[["test_ROC_AUC", "test_PR_AUC"]]
    table["base_rate"] = [y_train.iloc[te].mean() for _, te in cv.split(X_train)]
    table["PR_AUC / base"] = table["test_PR_AUC"] / table["base_rate"]
    table.index = range(1, len(table) + 1)
    table.index.name = "fold"
    return table.round(3)

from sklearn.ensemble import HistGradientBoostingClassifier

boost = HistGradientBoostingClassifier(early_stopping=False, random_state=0)

candidates = {
    "logreg + macro": build_pipeline(LogisticRegression(max_iter=1000), include_macro=True),
    "logreg, no macro": build_pipeline(LogisticRegression(max_iter=1000), include_macro=False),
    "boosting + macro": build_pipeline(boost, include_macro=True),
    "boosting, no macro": build_pipeline(boost, include_macro=False),
}

tables = {label: cv_table(mdl) for label, mdl in candidates.items()}
```

Let us compare the ROC AUC score in all four situations.

```python
pd.DataFrame({label: t["test_ROC_AUC"] for label, t in tables.items()})
```

Similarly, let us compare the increase in PR AUC score with respect to the base rate.

```python
pd.DataFrame({label: t["PR_AUC / base"] for label, t in tables.items()})
```

The results are somewhat surprising. Not only does adding gradient boosting not improve the results, but the model actually does worse than the simpler logistic regression. Furthermore, all cases show the aforementioned swings between folds, going down in fold 4 and then up in fold 5. Changing models didn't modify this behavior.

### No month

We try one more thing, namely removing the month.
This is because the month partly represents information about the period, so if we remove both the macro data and the month we will be left only with client and contact data.
If excluding the month makes the ROC AUC clearly better (or less erratic across folds), then including it was hurting under the time split.
If nothing improves, then it means that client and contact features carry little signal surviving in time in this data.

```python
boost_reg = HistGradientBoostingClassifier(
    max_depth=3, learning_rate=0.05, max_iter=150,
    min_samples_leaf=200, early_stopping=False, random_state=0)

more = {
    "logreg, no macro, no month": build_pipeline(
        LogisticRegression(max_iter=1000), include_macro=False, include_month=False),
    "boosting reg, no macro": build_pipeline(boost_reg, include_macro=False),
    "boosting reg, no macro, no month": build_pipeline(
        boost_reg, include_macro=False, include_month=False),
}
more_tables = {label: cv_table(m) for label, m in more.items()}

pd.DataFrame({label: t["test_ROC_AUC"] for label, t in more_tables.items()})
```

We can see that our guess was wrong: removing the month hurt the fit instead of helping. At least, this means that the real test will likely be better than the folds, because all months will have been seen.

## Summary

Let us summarize the different attempts in a table and give a brief overview of what we did.

```python
all_tables = {**tables, **more_tables}
all_tables["boosting reg + macro"] = cv_table(build_pipeline(boost_reg, include_macro=True))

summary = pd.DataFrame({
    label: {
        "ROC AUC mean": t["test_ROC_AUC"].mean(),
        "ROC AUC min": t["test_ROC_AUC"].min(),
        "ROC AUC max": t["test_ROC_AUC"].max(),
        "PR AUC / base mean": t["PR_AUC / base"].mean(),
    }
    for label, t in all_tables.items()
}).T.round(3)
summary
```

In this notebook we compared logistic regression and gradient boosting, with and without macroeconomic features (and in some cases also without `month`). We used a an expanding-window cross-validation with 5 folds inside the training data. The test set was not used.

Our main findings are:

- No model shows strong ranking skill in this setup. The mean ROC AUC score lies between 0.51 and 0.54 in all cases, and the PR AUC score is only about 1.0 to 1.2 times its base rate.
- Differences between folds (ROC AUC 0.42 to 0.66) are much larger than differences between models, so the period being predicted matters more than the model choice.
- Gradient boosting did not beat logistic regression, and excluding the macro columns made no clear difference.
- Removing `month` lowered the score, mainly in the last fold, where the scoring months also appear in training.
- Ordinary shuffled cross-validation gave ROC AUC 0.67 for logistic regression with macro features, against about 0.54 with the time-aware folds. Shuffling would have overstated the skill.

One caveat is that in these folds, 50% to 79% of scoring rows are from months the model has not seen in training. This means that the folds are harsher than the final test, where all months have been seen. We use the folds to compare models, not to predict test performance.

What we shall carry forward to the final evaluation is logistic regression and gradient boosting, each with and without macro features, all with `month`. Logistic regression is the main candidate as it matches or beats boosting here (on top of being easier to interpret).
