```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/bank.db")
```

```python
# Get a feel for the data (average age, call duration, ...)
df = pd.read_sql("""SELECT * FROM clients""", conn)
df.describe()
```

```python
# Check the subscription rate
query = """
    SELECT COUNT(*) AS n_clients,
        ROUND(100.0 * AVG(subscribed), 2) AS percentt_subscribed
    FROM clients;
"""
pd.read_sql(query, conn)
```

```python
# Find the conversion rate by job
df = pd.read_sql("""SELECT COUNT(job) FROM clients GROUP BY job""", conn)
```
